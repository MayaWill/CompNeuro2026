```python
import numpy as np
import matplotlib.pyplot as plt

a= np.linspace(0.01, 0.99, 100)

D = 1

c = np.sqrt(D) * ((1- (2*a)) / np.sqrt(a*(1-a)))

a_stall = (1/2)

#a_cross = (2 - np.sqrt(2)) / 4

passive_speed = 2 * np.sqrt(D)
```

```python
plt.plot(a, c)
plt.ylim(-10, 10)
plt.xlabel('a')
plt.ylabel('c')
plt.title('Wave speed c vs threshold a (D=1)')
plt.axhline(y=2, color='r', linestyle='--', label='passive speed = 2')
plt.axhline(y=0, linewidth=0.8)
plt.legend()
plt.plot((2 - np.sqrt(2))/4, 2, 'ko', markersize=6, label='crossing a* ≈ 0.146')
plt.show()
```
